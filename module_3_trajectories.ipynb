# Module 3 — Event trajectories

One pass over the raw data that turns both archives into a single event-level
table each: one row per event, with the cumulative intrinsic time `t`, the
cumulative novelty `D` and the cumulative explorer count `D_w`.

This cache is what the model is calibrated against: Module 4 scores every
candidate `N0` on it. `si_model_selection.ipynb` reads the same cache.

**Input** — the raw OpenAlex parquet files, and the augmented patent files
written by `module_1_patents.ipynb`.

**Output** — `outputs/trajectories/`, not versioned. Expect a long run: the
papers side is staged through a disk-backed SQLite database so that the result
is exact without holding the whole archive in memory.


```python
# Repository bootstrap.
from pathlib import Path
import sys

REPO_ROOT = Path.cwd().resolve()
while not (REPO_ROOT / "lib").is_dir() and REPO_ROOT != REPO_ROOT.parent:
    REPO_ROOT = REPO_ROOT.parent
if str(REPO_ROOT / "lib") not in sys.path:
    sys.path.insert(0, str(REPO_ROOT / "lib"))

DATA_DIR = REPO_ROOT / "data"                    # raw inputs, you provide these
OUTPUT_DIR = REPO_ROOT / "outputs"               # derived data, auto-generated
TRAJECTORY_DIR = OUTPUT_DIR / "trajectories"     # written here

# Papers without keywords cannot produce a combination novelty and are excluded
# from the event process. This is the convention of the revised manuscript.
filter_empty_keywords = True
```

## Imports and parameters


```python
from utils import OPENALEX_SUBFIELDS, build_all_trajectories

PATENTS_AUG_DIR = OUTPUT_DIR / "data_augmented_patents"
PAPERS_DIR = DATA_DIR / "papers_openalex"
STAGING_DIR = OUTPUT_DIR / "staging"

PAPERS_SUBFIELDS = OPENALEX_SUBFIELDS
```

## Build

The staging database is removed once the trajectory is written.


```python
build_all_trajectories(
    trajectory_dir=TRAJECTORY_DIR,
    patents_aug_dir=PATENTS_AUG_DIR,
    papers_dir=PAPERS_DIR,
    staging_dir=STAGING_DIR,
    papers_subfields=PAPERS_SUBFIELDS,
    cleanup_stage=True,
    filter_empty_keywords=filter_empty_keywords,
)
print("trajectories ->", TRAJECTORY_DIR)
```
