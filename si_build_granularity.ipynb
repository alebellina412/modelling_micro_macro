# SI — Granularity and contributor datasets

A second pass over the raw data, for the Supporting Information only. It
recounts the same events with novelty measured at three resolutions
(`D_single`, `D_pair`, `D_combo`) and creators counted under both the
focal-contributor and the all-contributor convention.

Run it once before `si_granularity.ipynb` and `si_contributors.ipynb`.

**Input** — the raw OpenAlex parquet files, and the augmented patent files
written by `module_1_patents.ipynb`.

**Output** — `outputs/granularity/`, not versioned.


```python
# Repository bootstrap.
from pathlib import Path
import sys

REPO_ROOT = Path.cwd().resolve()
while not (REPO_ROOT / "lib").is_dir() and REPO_ROOT != REPO_ROOT.parent:
    REPO_ROOT = REPO_ROOT.parent
if str(REPO_ROOT / "lib") not in sys.path:
    sys.path.insert(0, str(REPO_ROOT / "lib"))

DATA_DIR = REPO_ROOT / "data"                    # raw inputs, you provide these
OUTPUT_DIR = REPO_ROOT / "outputs"               # derived data, auto-generated
GRANULARITY_DIR = OUTPUT_DIR / "granularity"     # written here

# Papers without keywords cannot produce a combination novelty and are excluded
# from the event process. This is the convention of the revised manuscript.
filter_empty_keywords = True
```

## Imports and parameters


```python
import pandas as pd
from IPython.display import display

from analysis_utils import build_granularity_datasets
from utils import OPENALEX_SUBFIELDS

pd.set_option("display.max_columns", 80)

PATENTS_AUG_DIR = OUTPUT_DIR / "data_augmented_patents"
PAPERS_DIR = DATA_DIR / "papers_openalex"

START_YEAR = 1980
END_YEAR = 2020
PAPERS_SUBFIELDS = OPENALEX_SUBFIELDS
```

## Build


```python
bundles = build_granularity_datasets(
    output_dir=GRANULARITY_DIR,
    patents_aug_dir=PATENTS_AUG_DIR,
    papers_dir=PAPERS_DIR,
    start_year=START_YEAR,
    end_year=END_YEAR,
    subfields=PAPERS_SUBFIELDS,
    filter_empty_keywords=filter_empty_keywords,
)

display(pd.DataFrame([{
    "dataset": name,
    "events": str(bundle.events_path.relative_to(REPO_ROOT)),
    "annual": str(bundle.annual_path.relative_to(REPO_ROOT)),
} for name, bundle in bundles.items()]))
```
