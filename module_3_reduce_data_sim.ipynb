## Create reduced simulation files (patents)

This cell reads the model simulation output for patents and creates the reduced log/linear time series plus the frequency table saved in `outputs/`.


```python
from pathlib import Path
import numpy as np

T = 41
nu = 1.0
rho = 1.0
N0 = 120000
a = 0.0545
b = 1.33
w0 = 269781

model_prefix = "model_T=%d_nu=%.6g_rho=%.6g_N0=%d_a=%.6g_b=%.6g_w0=%.6g" % (
    T, nu, rho, N0, a, b, w0
)
model_path = Path("data_simulations") / f"{model_prefix}.dat"

with open(model_path, "r") as f:
    lines = f.readlines()

try:
    data = np.loadtxt(lines)
except ValueError:
    data = np.loadtxt(lines[:-1])

t = data[:, 1]
D = data[:, 2]
w = data[:, 5]

print(f"Total time: {T - 1}")
print(f"Total number of events (t): {int(t[-1])}")
print(f"Total number of explorers (w): {int(np.max(w))}")
print(f"Total novelties (D): {int(np.max(D))}")
```

```python
import numpy as np
import pandas as pd
from collections import Counter
from pathlib import Path

# === Model parameters (must match the Module 3 run) ===
T = 41
nu = 1.0
rho = 1.0
N0 = 120000

a = 0.0545
b = 1.33
w0 = 269781

model_dir = Path("data_simulations")
output_dir = Path("outputs")
output_dir.mkdir(parents=True, exist_ok=True)
model_out_dir = output_dir

model_prefix = "model_T=%d_nu=%.6g_rho=%.6g_N0=%d_a=%.6g_b=%.6g_w0=%.6g" % (
    T, nu, rho, N0, a, b, w0
)
model_path = model_dir / f"{model_prefix}.dat"

# Read lines as raw text
with open(model_path, "r") as f:
    lines = f.readlines()

# Try to load; if it fails, drop the last corrupted line
try:
    data = np.loadtxt(lines)
except ValueError as e:
    print(f"Warning '{e}', dropping last corrupted line.")
    data = np.loadtxt(lines[:-1])

# Columns: tau, t, D, estr, num_estr, w

tau = data[:, 0]
t = data[:, 1]
D = data[:, 2]
estr = data[:, 3]
num_estr = data[:, 4]
w = data[:, 5]

# Frequency of extracted elements
counts = Counter(estr)
freq_D_sim = np.array(sorted(counts.values(), reverse=True))
sim_ranks = np.arange(1, len(freq_D_sim) + 1)

# Extract only N = 10**4 log and linear indices, then save as before.
N = len(tau)
n_points = min(10**4, N)
log_indices = np.unique(np.logspace(0, np.log10(N), n_points, dtype=int) - 1)
lin_indices = np.linspace(0, N - 1, n_points, dtype=int)

# Log samples

tau_log = tau[log_indices]
t_log = t[log_indices]
D_log = D[log_indices]
num_estr_log = num_estr[log_indices]
w_log = w[log_indices]

# Linear samples

tau_lin = tau[lin_indices]
t_lin = t[lin_indices]
D_lin = D[lin_indices]
num_estr_lin = num_estr[lin_indices]
w_lin = w[lin_indices]

# Save reduced time series

df_log_sim = pd.DataFrame({
    "tau": tau_log,
    "t": t_log,
    "D": D_log,
    "num_estr": num_estr_log,
    "w": w_log,
})

df_lin_sim = pd.DataFrame({
    "tau": tau_lin,
    "t": t_lin,
    "D": D_lin,
    "num_estr": num_estr_lin,
    "w": w_lin,
})

df_log_sim.to_csv(model_out_dir / f"{model_prefix}_log.csv", index=False)
df_lin_sim.to_csv(model_out_dir / f"{model_prefix}_lin.csv", index=False)

# Save frequency table

df_freq = pd.DataFrame({
    "rank": sim_ranks,
    "frequency": freq_D_sim,
})

df_freq.to_csv(model_out_dir / f"{model_prefix}_freq.csv", index=False)
```
