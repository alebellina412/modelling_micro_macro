# Module 5 — Reduce the simulation output

The C model writes one row per extraction, tens of millions of rows per run:
too large to plot directly. This notebook subsamples each of the two published
runs onto a log and a linear grid and extracts the rank-frequency table of the
elements, writing the CSVs that Module 6 compares against the data.

Input is the `.dat` produced by `module_4_n0_search.ipynb` at the selected
`N0`, under `sim_runs/<dataset>/data_simulations/`.


```python
# Repository bootstrap.
# The notebooks are meant to be run from the repository root; the walk keeps
# them working when a front-end starts the kernel in a subdirectory.
from pathlib import Path
import sys

REPO_ROOT = Path.cwd().resolve()
while not (REPO_ROOT / "lib").is_dir() and REPO_ROOT != REPO_ROOT.parent:
    REPO_ROOT = REPO_ROOT.parent
if str(REPO_ROOT / "lib") not in sys.path:
    sys.path.insert(0, str(REPO_ROOT / "lib"))

OUTPUT_DIR = REPO_ROOT / "outputs"     # derived data, auto-generated
SIM_RUNS_DIR = REPO_ROOT / "sim_runs"  # model runs (N0 scan and selected run)

from analysis_utils import reduce_simulation_run
```

## Patents

Manuscript calibration for the patents-like run. Writes
`model_T=41_..._N0=100500_..._{log,lin,freq}.csv` into `outputs/`.


```python
T, nu, rho = 41, 1.0, 1.0
N0, a, b, w0 = 100500, 0.0545, 1.33, 269781

model_prefix = "model_T=%d_nu=%.6g_rho=%.6g_N0=%d_a=%.6g_b=%.6g_w0=%.6g" % (
    T, nu, rho, N0, a, b, w0
)
dat_path = SIM_RUNS_DIR / "patents" / "data_simulations" / f"{model_prefix}.dat"

summary = reduce_simulation_run(dat_path, OUTPUT_DIR, model_prefix)
print(f"Total time: {T - 1}")
print(f"Total number of events (t): {summary['events']}")
print(f"Total number of explorers (w): {summary['explorers']}")
print(f"Total novelties (D): {summary['novelties']}")
```

```python
T, nu, rho = 41, 1.0, 1.0
N0, a, b, w0 = 49000, 0.167, 0.71, 212313

model_prefix = "model_T=%d_nu=%.6g_rho=%.6g_N0=%d_a=%.6g_b=%.6g_w0=%.6g" % (
    T, nu, rho, N0, a, b, w0
)
dat_path = SIM_RUNS_DIR / "papers" / "data_simulations" / f"{model_prefix}.dat"

summary = reduce_simulation_run(dat_path, OUTPUT_DIR, model_prefix)
print(f"Total time: {T - 1}")
print(f"Total number of events (t): {summary['events']}")
print(f"Total number of explorers (w): {summary['explorers']}")
print(f"Total novelties (D): {summary['novelties']}")
```
